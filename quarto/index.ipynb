::: {.home-lead}
Collaborative competitions are now a standard way to evaluate scientific and
technological progress: organizers define a task, participants submit
predictions for a held-out dataset, and the systems are ranked by
performance. The last step -- ranking -- is harder than it looks. **CompStats**,
developed by [INGEOTEC](https://github.com/INGEOTEC), gives organizers and
participants the statistical tools to do it rigorously.
:::

Traditional statistical tests are routinely used to decide whether one
system outperforms another, but they often fail to capture decisive
differences between systems' performance, particularly when many
participants must be compared and ranked at once. Nava-Muñoz, Graff and
Escalante address this in [*Analysis of Systems' Performance in Natural
Language Processing Competitions*](https://arxiv.org/abs/2403.04693), which
introduces a bootstrap-based evaluation methodology and validates it on eight
NLP competitions spanning classification and regression tasks. `CompStats` is
the implementation of that methodology: it draws paired bootstrap resamples
of the held-out set, evaluates every system's predictions on the same
resamples, and derives standard errors, confidence intervals, and p-values
for the difference against the best system directly from the resulting
distributions -- without the assumptions a parametric test would require.

::: {.pillars}
::: {.pillar}
### Off-the-shelf comparisons

Follows the `sklearn.metrics` convention -- a score function takes
`(y_true, y_pred)` -- so evaluating a new system is a one-line change.
:::
::: {.pillar}
### Confidence intervals

Standard error and percentile confidence intervals for every metric come
from the bootstrap distribution, with no extra configuration.
:::
::: {.pillar}
### Paired significance testing

Every system is resampled with the same bootstrap indices, so the p-value
of a difference against the best system is valid by construction.
:::
:::

## Getting started

This site is organized as one notebook per topic: [scikit-learn
Users](pages/sklearn_users.ipynb) walks through wrapping predictions from
several classifiers in a `Perf` instance, [Competition
Organizers](pages/competition_organizers.ipynb) shows how to rank
participants and report confidence intervals against a gold standard, and
[Multiple Testing](pages/multiple_testing.ipynb) covers comparing several
measures at once with correction for multiple comparisons. Every page,
including this one, is a real Jupyter notebook that can also be run outside
the site.
